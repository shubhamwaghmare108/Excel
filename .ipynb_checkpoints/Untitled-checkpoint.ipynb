```python
lessthanten = []  # List to store column names with < 10 unique values

for col in titanic.columns:
    if titanic[col].nunique() < 10:
        lessthanten.append(col)  

for col in lessthanten:
    print(col, titanic[col].unique())  # Print column name and unique values
```

```python
# Plot count plots for these columns
plt.figure(figsize=(15, 10))

for i, col in enumerate(lessthanten, 1):
    plt.subplot(3, 3, i)  # Adjust the grid layout if needed
    ax = sns.countplot(data=titanic, x=col)

    # Add data labels
    for p in ax.patches:
        ax.text(p.get_x() + p.get_width() / 2,  # X position
                p.get_height() + 1,             # Y position (slightly above bar)
                int(p.get_height()),            # Label (count value)
                ha='center', va='bottom', fontsize=10, fontweight='bold')

    plt.title(f'Count Plot of {col}')
    plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Example: Show average fare for each passenger class
plt.figure(figsize=(8, 6))
ax = sns.barplot(data=titanic, x="Pclass", y="Fare", estimator=sum)  # Change estimator as needed

# Add data labels
for p in ax.patches:
    ax.text(p.get_x() + p.get_width() / 2,  # X position (center of bar)
            p.get_height() + 5,             # Y position (slightly above bar)
            f'{p.get_height():.0f}',        # Label (rounded value)
            ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.title("Total Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Total Fare")
plt.xticks(rotation=0)
plt.show()
```
